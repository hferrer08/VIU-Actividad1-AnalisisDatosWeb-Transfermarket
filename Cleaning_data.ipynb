# Actividad 1 - Análisis de datos web

## Dataset seleccionado

Football Data from Transfermarkt.

## Objetivo

Analizar información de jugadores, clubes y competiciones para construir un dashboard interactivo en Power BI que permita identificar tendencias relacionadas con el valor de mercado de los jugadores y las características de los clubes.

```python
import pandas as pd

players = pd.read_csv("/players.csv")
clubs = pd.read_csv("/clubs.csv")
competitions = pd.read_csv("/competitions.csv")
valuations = pd.read_csv("/player_valuations.csv")

print("Archivos cargados correctamente")
```

```python
# Revisamos el tamaño de cada DataFrame.
# El resultado se muestra como (filas, columnas), similar a revisar cuántos registros
# y campos tiene cada tabla.
print("players:", players.shape)
print("clubs:", clubs.shape)
print("competitions:", competitions.shape)
print("valuations:", valuations.shape)
```

```python
# Visualizamos las primeras filas de la tabla de jugadores.
players.head()
```

```python
# Revisamos la estructura de la tabla de jugadores
players.info()
```

```python
# Obtenemos el listado de columnas de cada DataFrame.
# Campos y relaciones entre tablas
print("Columnas players:")
print(players.columns.tolist())

print("\nColumnas clubs:")
print(clubs.columns.tolist())

print("\nColumnas competitions:")
print(competitions.columns.tolist())

print("\nColumnas valuations:")
print(valuations.columns.tolist())
```

# Exploración inicial del dataset

En esta sección se cargan los archivos CSV y se analiza su estructura inicial.

Se identifican las tablas principales del modelo de datos:

- Players: información de jugadores.
- Clubs: información de clubes.
- Competitions: información de competiciones.
- Player Valuations: histórico de valor de mercado de los jugadores.

Posteriormente se analizarán los tipos de datos, valores nulos y relaciones entre tablas para preparar la información utilizada en el dashboard de Power BI.

```python
# Analizamos la cantidad de valores nulos presentes en cada tabla
print("Nulos en players:")
print(players.isnull().sum())

print("\nNulos en clubs:")
print(clubs.isnull().sum())

print("\nNulos en competitions:")
print(competitions.isnull().sum())

print("\nNulos en valuations:")
print(valuations.isnull().sum())
```

```python
# Verificamos si existen registros duplicados en cada tabla
print("Duplicados en players:", players.duplicated().sum())
print("Duplicados en clubs:", clubs.duplicated().sum())
print("Duplicados en competitions:", competitions.duplicated().sum())
print("Duplicados en valuations:", valuations.duplicated().sum())
```

```python
# Seleccionamos solo las columnas necesarias para el análisis.
# Esto permite simplificar el modelo de datos y eliminar campos que no aportan valor al dashboard.

players_clean = players[[
    "player_id",
    "name",
    "last_season",
    "current_club_id",
    "country_of_birth",
    "country_of_citizenship",
    "date_of_birth",
    "position",
    "sub_position",
    "foot",
    "height_in_cm",
    "current_club_domestic_competition_id",
    "current_club_name",
    "market_value_in_eur",
    "highest_market_value_in_eur"
]].copy()

clubs_clean = clubs[[
    "club_id",
    "name",
    "domestic_competition_id",
    "squad_size",
    "average_age",
    "foreigners_number",
    "foreigners_percentage",
    "national_team_players",
    "stadium_name",
    "stadium_seats"
]].copy()

competitions_clean = competitions[[
    "competition_id",
    "name",
    "type",
    "country_name",
    "confederation",
    "total_clubs"
]].copy()

valuations_clean = valuations[[
    "player_id",
    "date",
    "market_value_in_eur",
    "current_club_id",
    "current_club_name"
]].copy()

print("Columnas seleccionadas correctamente")
```

```python
# Revisamos el tamaño de las tablas limpias después de seleccionar las columnas relevantes.

print("players_clean:", players_clean.shape)
print("clubs_clean:", clubs_clean.shape)
print("competitions_clean:", competitions_clean.shape)
print("valuations_clean:", valuations_clean.shape)
```

```python
# Visualizamos los primeros registros de las tablas limpias.
# Esto permite validar rápidamente la estructura y el contenido antes de realizar relaciones entre tablas.

print("PLAYERS")
display(players_clean.head())

print("CLUBS")
display(clubs_clean.head())

print("COMPETITIONS")
display(competitions_clean.head())

print("VALUATIONS")
display(valuations_clean.head())
```

```python
# Validamos relaciones entre tablas
# Confirmamos cuántos jugadores tienen un club existente en la tabla clubs
# y cuántas valoraciones tienen un jugador existente en la tabla players.

print("Total jugadores:", len(players_clean))
print("Jugadores con club existente en clubs:")
print(players_clean["current_club_id"].isin(clubs_clean["club_id"]).sum())

print("\nTotal valoraciones:", len(valuations_clean))
print("Valoraciones con jugador existente en players:")
print(valuations_clean["player_id"].isin(players_clean["player_id"]).sum())

print("\nClubes con competición existente:")
print(clubs_clean["domestic_competition_id"].isin(competitions_clean["competition_id"]).sum())
```

```python
# Construimos una tabla analítica unificando jugadores,
# clubes y competiciones mediante sus relaciones.

players_dashboard = (
    players_clean
    .merge(
        clubs_clean,
        left_on="current_club_id",
        right_on="club_id",
        how="left",
        suffixes=("_player", "_club")
    )
    .merge(
        competitions_clean,
        left_on="domestic_competition_id",
        right_on="competition_id",
        how="left"
    )
)

print("Tabla dashboard creada correctamente")
print(players_dashboard.shape)
```

```python
# Visualizamos los primeros registros de la tabla integrada.

players_dashboard.head()
```

```python
# Renombramos algunas columnas para que sean más claras en Power BI
# Convertimos la fecha de nacimiento a tipo fecha y calculamos la edad aproximada del jugador

players_dashboard = players_dashboard.rename(columns={
    "name_player": "player_name",
    "name_club": "club_name",
    "name": "competition_name",
    "country_name": "competition_country"
})

players_dashboard["date_of_birth"] = pd.to_datetime(
    players_dashboard["date_of_birth"],
    errors="coerce"
)

players_dashboard["age"] = 2026 - players_dashboard["date_of_birth"].dt.year

print("Columnas renombradas y edad calculada correctamente")
```

```python
# Revisamos las columnas finales disponibles para el dashboard.

players_dashboard.columns.tolist()
```

```python
# Seleccionamos las columnas finales que se exportarán para Power BI.

dashboard_final = players_dashboard[[
    "player_id",
    "player_name",
    "age",
    "last_season",
    "country_of_birth",
    "country_of_citizenship",
    "position",
    "sub_position",
    "foot",
    "height_in_cm",
    "current_club_id",
    "current_club_name",
    "market_value_in_eur",
    "highest_market_value_in_eur",
    "club_id",
    "squad_size",
    "average_age",
    "foreigners_number",
    "foreigners_percentage",
    "national_team_players",
    "stadium_name",
    "stadium_seats",
    "competition_id",
    "competition_name",
    "type",
    "competition_country",
    "confederation",
    "total_clubs"
]].copy()

dashboard_final.head()
```

```python
# Exportamos el dataset final limpio en formato CSV.
# Este archivo será cargado posteriormente en Power BI para construir el dashboard.

dashboard_final.to_csv("/dashboard_transfermarkt.csv", index=False)

print("Archivo dashboard_transfermarkt.csv generado correctamente")
```

```python
# Revisamos el tamaño final del dataset exportado para Power BI.
print("Dataset final:", dashboard_final.shape)
```

```python
# Revisamos si el dataset final conserva valores nulos después de la preparación.
dashboard_final.isnull().sum()
```

```python
# Reemplazamos valores nulos de texto por "No informado"
# para que Power BI muestre una categoría clara en los filtros y gráficos.

text_columns = [
    "country_of_birth",
    "country_of_citizenship",
    "sub_position",
    "foot",
    "current_club_name",
    "stadium_name",
    "competition_name",
    "type",
    "competition_country",
    "confederation"
]

dashboard_final[text_columns] = dashboard_final[text_columns].fillna("No informado")

# Reemplazamos valores nulos numéricos por 0 cuando representan ausencia de dato.
numeric_columns = [
    "market_value_in_eur",
    "highest_market_value_in_eur",
    "height_in_cm",
    "squad_size",
    "average_age",
    "foreigners_number",
    "foreigners_percentage",
    "national_team_players",
    "stadium_seats",
    "total_clubs"
]

dashboard_final[numeric_columns] = dashboard_final[numeric_columns].fillna(0)

print("Valores nulos tratados correctamente")
```
